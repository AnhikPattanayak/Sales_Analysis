# Task 4 : Data Cleaning

```python
# making an copy of data

temp = df.copy()
```

```python
# drop fully-empty columns

temp.drop(columns = {'Status', 'unnamed1'}, inplace = True)
```

```python
# Checking for null values

temp.isnull().sum()
```

```python
print(f'Null values befor handling:\n{temp.isnull().sum()[temp.isnull().sum()>0]}')
```

```python
# handling null values

temp[temp['Amount'].isnull()]
```

```python
# Replacing null values in amount with the average amt. of Product_Category & Zone group

temp['Amount'] = temp['Amount'].fillna(temp.groupby(['Product_Category','Zone'])['Amount'].transform('mean'))
```

```python
print(f'Number of duplicate entries : {temp.duplicated().sum()}')
```

```python
# drop duplicates

temp.drop_duplicates(inplace = True)
```

```python
print(f'After Cleaning Dataset : {temp.shape[0]:,} rows x {temp.shape[1]} columns')
```

```python
# change dtype of Amount

temp['Amount'] = temp['Amount'].astype('int')
```

```python
temp['Marital_Status'] = temp['Marital_Status'].map({0: 'Unmarried', 1: 'Married'})
```

```python
# snakecasing the clolumn names

temp.columns = temp.columns.str.lower()
temp.columns = temp.columns.str.replace(' ','_')
```

# Task 5 : Exploratory Data Analysis

```python
# Feature Engg.
# 1. Average spend per order

temp['avg_order_value'] = (temp['amount']/temp['orders']).round(2)
```

```python
# 2. Number of purchases by each person

purchase_freq = temp.groupby('user_id').size().reset_index(name = 'purchase_count')
temp = temp.merge(purchase_freq, how = 'left', on = 'user_id')
```

```python
# 3. Loyalty levels

loyalty = ['One-time', 'Occasional', 'Regular', 'Loyal']
temp['loyalty_tiers'] = pd.cut(temp['purchase_count'], bins = [0, 1, 3, 6, 100], labels = loyalty)
temp.head()
```

```python
zone_rev = temp.groupby('zone')['amount'].sum().reset_index()
zone_rev['zone_pct'] = (zone_rev['amount']/zone_rev['amount'].sum() * 100).round(1)
zone_rev.sort_values('amount', ascending = False, inplace = True)

fig, ax = plt.subplots(1, 2, figsize = (14,5))
for single_ax in ax.flatten():
    # Keep only bottom (x) and left (y) axis lines
    single_ax.spines['top'].set_visible(False)
    single_ax.spines['right'].set_visible(False)
    
    # Gridlines
    single_ax.grid(True, axis = 'both', linestyle = 'solid', alpha = 0.5, color = 'gray')
    single_ax.set_axisbelow(True)

# Bar chart : revenue per zone

bars = ax[0].barh(zone_rev['zone'], zone_rev['amount'])
ax[0].bar_label(bars, labels = [f'₹{v/1e6:.1f}M' for v in zone_rev['amount']], padding=5)
ax[0].set_title('Revenue by Zone', fontweight = 'bold')
ax[0].set_xlabel('Total Revenue(₹)')
ax[0].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{x/1e6:.0f}M'))
ax[0].set_ylabel('Zones')

# Pie chart : revenue per zone percentage

ax[1].pie(zone_rev['zone_pct'], autopct = '%0.1f%%', labels = zone_rev['zone'])
ax[1].set_title('Zone Revenue share(%)', fontweight = 'bold')

plt.suptitle('Regional Revenue Distribution', fontsize = 14, fontweight = 'bold', y = 1.01)
plt.tight_layout()
plt.show()
```

*Central zone alone contributes ~39% of total revenue with around ₹41.5M — driven by Uttar Pradesh, Delhi, and Madhya Pradesh. Eastern zone is significantly underrepresented at ~7%, signalling an untapped market.*

```python
# State-level breakdown within each Zone

state_zone = temp.groupby(['zone','state'])['amount'].sum().reset_index()
state_zone = state_zone.sort_values(['zone','amount'], ascending = [True, False])

fig, ax = plt.subplots(1, 5, figsize = (22, 5))
for single_ax in ax.flatten():
    # Keep only bottom (x) and left (y) axis lines
    single_ax.spines['top'].set_visible(False)
    single_ax.spines['right'].set_visible(False)
    
    # Gridlines
    single_ax.grid(True, axis = 'both', linestyle = 'solid', alpha = 0.5, color = 'gray')
    single_ax.set_axisbelow(True)
zones = sorted(state_zone['zone'].unique())

for i, z in enumerate(zones):
    data = state_zone[state_zone['zone'] == z]
    bars = ax[i].barh(data['state'], data['amount'])
    ax[i].set_title(f'{z} Zone', fontweight = 'bold')
    ax[i].set_xlabel('Revenue(₹)')
    ax[i].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{x/1e6:.0f}M'))
    ax[i].invert_yaxis()

    for bar in bars:
        ax[i].text(bar.get_width() * 1.02, bar.get_y() + bar.get_height()/2,
                     f'₹{bar.get_width()/1e6:.1f}M', va = 'center', fontsize = 10)

plt.suptitle('Top States by Revenue Within Each Zone', fontsize = 14, fontweight = 'bold', y = 1.01)
plt.tight_layout()
plt.show()
```

*Within Central zone, UP alone contributes ₹19M. In the South, Karnataka leads with value of ₹13.5M. Northern zone revenue is thin and spread across 4 states & Eastern zone is showing less involvement.*

```python
fig, ax = plt.subplots(1, 2, figsize = (14, 5))

for single_ax in ax.flatten():
    # Keep only bottom (x) and left (y) axis lines
    single_ax.spines['top'].set_visible(False)
    single_ax.spines['right'].set_visible(False)
    
    # Gridlines
    single_ax.grid(True, axis = 'both', linestyle = 'solid', alpha = 0.5, color = 'gray')
    single_ax.set_axisbelow(True)
    
# Gender × Marital Status: avg spend per buyer

gm = temp.groupby(['marital_status', 'gender'])['amount'].mean().reset_index()
gm = gm.sort_values('amount', ascending = False)

g1 = sns.barplot(data = gm, x = 'marital_status', y = 'amount', hue = 'gender', ax = ax[0])
ax[0].set_title('Avg Spend per Buyer\nby Marital Status × Gender', fontweight = 'bold')
ax[0].set_xlabel('Marital Status')
ax[0].set_ylabel('Avg Spend(₹)')
ax[0].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{x:,.0f}'))
for bars in g1.containers:
    g1.bar_label(bars)

# Age Group × Gender: buyer count heatmap-style grouped bar

ag = temp.groupby(['age_group', 'gender']).size().reset_index(name = 'count')
age_order = ['0-17', '18-25', '26-35', '36-45', '46-50', '51-55', '55+']
ag['age_group'] = pd.Categorical(ag['age_group'], categories = age_order, ordered = True)
ag.sort_values('age_group')

g2 = sns.barplot(data = ag, x = 'age_group', y = 'count', hue = 'gender', ax = ax[1])
ax[1].set_title("Buyer Count by Age Group × Gender", fontweight="bold")
ax[1].set_ylabel('Number of Buyers')
ax[1].set_xlabel('Age Group')
for bars in g2.containers:
    g2.bar_label(bars)

plt.tight_layout()
plt.show()
```

*Unmarried has highest avg spend per transaction. The 26-35 age_group dominates in volume.*

```python
# Occupation: avg order value (fairer metric than total, avoids headcount bias)

occ = temp.groupby(['occupation']).agg(total_revenue = ('amount','sum'), buyer_count = ('user_id','nunique'), avg_order_value = ('avg_order_value','mean')).reset_index().sort_values('avg_order_value', ascending = False)

fig, ax = plt.subplots(1, 2, figsize = (16, 5))
for single_ax in ax.flatten():
    # Keep only bottom (x) and left (y) axis lines
    single_ax.spines['top'].set_visible(False)
    single_ax.spines['right'].set_visible(False)
    
    # Gridlines
    single_ax.grid(True, axis = 'both', linestyle = 'solid', alpha = 0.5, color = 'gray')
    single_ax.set_axisbelow(True)
    
bh1 = ax[0].barh(occ['occupation'], occ['avg_order_value'])
ax[0].set_title('Occupations Ranked by\n Avg Order Value', fontweight = 'bold')
ax[0].invert_yaxis()
ax[0].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{x:,.0f}'))
ax[0].bar_label(bh1 , labels = [f'₹{v:,.0f}' for v in occ['avg_order_value']], padding = 4, fontsize = 10)
ax[0].set_ylabel('Occupations')
ax[0].set_xlabel('Avg Spend per Order(₹)')

# Right: Total revenue

occ2 = occ.sort_values("total_revenue", ascending = False)
bh2 = ax[1].barh(occ2['occupation'], occ2['total_revenue'])
ax[1].set_title('Occupations Ranked by\n Total Revenue', fontweight = 'bold')
ax[1].invert_yaxis()
ax[1].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{x/1e6:.1f}M'))
ax[1].bar_label(bh2, labels = [f'₹{v/1e6:.1f}M' for v in occ2['total_revenue']], padding = 4, fontsize = 10)
ax[1].set_ylabel('Occupations')
ax[1].set_xlabel('Total Revenue(₹)')

plt.tight_layout()
plt.show()
```

*Govt and Chemical employees have the highest avg order value. IT Sector dominates total revenue. Targeting Govt/Chemical buyers with premium products could improve margins.*

```python
# Avg order value by Age Group (not just buyer count)

age_aov = temp.groupby('age_group')['avg_order_value'].mean().reset_index()
age_order = ['0-17', '18-25', '26-35', '36-45', '46-50', '51-55', '55+']
age_aov['age_group'] = pd.Categorical(age_aov['age_group'], categories = age_order, ordered = True)
age_aov = age_aov.sort_values('age_group')

fig, ax = plt.subplots(1, 2, figsize = (15, 5))
for single_ax in ax.flatten():
    # Keep only bottom (x) and left (y) axis lines
    single_ax.spines['top'].set_visible(False)
    single_ax.spines['right'].set_visible(False)
    
    # Gridlines
    single_ax.grid(True, axis = 'both', linestyle = 'solid', alpha = 0.5, color = 'gray')
    single_ax.set_axisbelow(True)
# AOV by Age Group
b1 = ax[0].bar(age_aov['age_group'], age_aov['avg_order_value'])
ax[0].set_title('Avg Spend per Order by Age Group', fontweight = 'bold')
ax[0].bar_label(b1 , labels = [f'₹{v:,.0f}' for v in age_aov['avg_order_value']], padding = 4, fontsize = 10)
ax[0].set_ylabel('Avg Order Value(₹)')
ax[0].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{x:,.0f}'))
ax[0].set_xlabel('Age Group')

# Spend distribution (KDE)
ax[1].hist(temp['amount'], bins = 40, edgecolor = 'white')
ax[1].axvline(temp['amount'].mean(), color = 'red', linestyle = 'dashed', linewidth = 2, label = f'Mean ₹{temp['amount'].mean():,.0f}')
ax[1].axvline(temp['amount'].median(), color = 'black', linestyle = 'dashdot', linewidth = 2, label = f'Median ₹{temp['amount'].median():,.0f}')
ax[1].set_title('Distribution of Transaction Amount', fontweight = 'bold')
ax[1].set_xlabel('Amount (₹)')
ax[1].set_ylabel('Frequency')
ax[1].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{x/1e3:.0f}K'))
ax[1].legend()

plt.tight_layout()
plt.show()
```

*46-50 and 51-55 age group have higher avg order values. This is a key finding for upselling strategies: older buyers are fewer but higher-value per transaction.*

```python
# Loyalty tier spending analysis

loyalty_spend = temp.groupby('loyalty_tiers').agg(unique_buyers = ('user_id', 'nunique'), total_revenue = ('amount', 'sum'), avg_spend = ('amount', 'mean')).reset_index()
loyalty_spend['revenue_pct'] = (loyalty_spend['total_revenue']/loyalty_spend['total_revenue'].sum() * 100).round(1)

fig, ax = plt.subplots(1, 3, figsize = (16, 5))
for single_ax in ax.flatten():
    # Keep only bottom (x) and left (y) axis lines
    single_ax.spines['top'].set_visible(False)
    single_ax.spines['right'].set_visible(False)
    
    # Gridlines
    single_ax.grid(True, axis = 'both', linestyle = 'solid', alpha = 0.5, color = 'gray')
    single_ax.set_axisbelow(True)

# Buyers count
l0 = ax[0].bar(loyalty_spend['loyalty_tiers'], loyalty_spend['unique_buyers'])
ax[0].set_title('Buyer Count by Tier', fontweight = 'bold')
ax[0].set_xlabel('Loyalty level')
ax[0].set_ylabel('Unique Buyers')
ax[0].bar_label(l0 , labels = [f'{v}' for v in loyalty_spend['unique_buyers']], padding = 4, fontsize = 10)


# Revenue %
l1 = ax[1].bar(loyalty_spend['loyalty_tiers'], loyalty_spend['revenue_pct'])
ax[1].set_title('Revenue Share by Tier (%)', fontweight = 'bold')
ax[1].set_xlabel('Loyalty level')
ax[1].set_ylabel('Revenue share')
ax[1].bar_label(l1 , labels = [f'{v}%' for v in loyalty_spend['revenue_pct']], padding = 4, fontsize = 10)


# Avg spend per transaction
l2 = ax[2].bar(loyalty_spend['loyalty_tiers'], loyalty_spend['avg_spend'])
ax[2].set_title('Avg Transaction Value by Tier', fontweight = 'bold')
ax[2].set_xlabel('Loyalty level')
ax[2].set_ylabel('Average Amount(₹)')
ax[2].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{x:,.0f}'))
ax[2].bar_label(l2 , labels = [f'₹{v:,.0f}' for v in loyalty_spend['avg_spend']], padding = 4, fontsize = 10)


plt.suptitle('Buyer Loyalty Tier Analysis', fontsize = 13, fontweight = 'bold')
plt.tight_layout()
plt.show()
```

*One-time buyers are the largest group (1,400 unique buyers) but Loyal buyers (6+ purchases) drive disproportionate revenue per transaction. Retention strategies for Occasional buyers converting them to Regular could significantly boost total revenue.*

```python
cat_group = temp.groupby('product_category').agg( total_orders = ('orders', 'sum'), total_revenue = ('amount', 'sum'), avg_price = ('avg_order_value', 'mean')).reset_index()
cat_top = cat_group.sort_values('total_revenue', ascending = False).head(10)

# Revenue
plt.figure(figsize = (12,5))
ax = plt.gca()
# Keep only bottom (x) and left (y) axis lines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Gridlines
ax.grid(True, axis = 'both', linestyle = 'solid', alpha = 0.5, color = 'gray')
ax.set_axisbelow(True)

g = plt.barh(cat_top['product_category'], cat_top['total_revenue'])
plt.title('Top 10 Categories by Revenue', fontweight = 'bold')
plt.xlabel('Total Revenue (₹)')
plt.bar_label(g, labels = [f'₹{v/1e6:.1f}M' for v in cat_top['total_revenue']], padding = 4)
plt.gca().xaxis.set_major_formatter(mticker.FuncFormatter(lambda x,_: f'₹{x/1e6:.1f}M'))

plt.tight_layout()
plt.show()
```

*Category-wise Food provides the highest revenue with ₹34.0M.*

```python
# Zone × Top Category heatmap
zone_cat = temp.groupby(['zone', 'product_category'])['amount'].sum().reset_index()
zone_cat_pivot = zone_cat.pivot(index = 'zone', columns = 'product_category', values = 'amount')

# Keep only top 10 categories by total revenue
top_cats = cat_group.sort_values('total_revenue', ascending = False)['product_category'].head(10).tolist()
zone_cat_pivot = zone_cat_pivot[top_cats]

plt.figure(figsize = (16, 6))
sns.heatmap(zone_cat_pivot/1e6, annot = True, fmt = '.1f', cmap = 'YlOrBr', linewidths = 0.5, cbar_kws = {'label': 'Revenue (₹M)'})
plt.title('Zone × Product Category Revenue Heatmap (₹ Millions)', fontweight = 'bold')
plt.xlabel('Product Category')
plt.ylabel('Zone')
plt.xticks(rotation = 30)
plt.tight_layout()
plt.show()
```

*Central zone dominates Food (₹13.7M) and Footwear. Northern zone is noticeably weak across all categories — especially Food and Electronics. Eastern zone has almost no presence in high-revenue categories, suggesting distribution gaps rather than demand gaps.*
